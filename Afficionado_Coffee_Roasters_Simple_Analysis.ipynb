# Afficionado Coffee Roasters – Sales Trend Analysis

Simple Jupyter Notebook for the Unified Mentor project.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```
df = pd.read_csv("Afficionado Coffee Roasters.xlsx - Transactions.csv")
df.head()
```

```
print("Shape:", df.shape)
df.info()
```

```
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```
df["Revenue"] = df["transaction_qty"] * df["unit_price"]
df.head()
```

```
df["transaction_time"] = pd.to_datetime(
    df["transaction_time"],
    format="%H:%M:%S"
)
df["Hour"] = df["transaction_time"].dt.hour
df.head()
```

```
def time_bucket(hour):
    if 6 <= hour <= 11:
        return "Morning"
    elif 12 <= hour <= 16:
        return "Afternoon"
    elif 17 <= hour <= 21:
        return "Evening"
    else:
        return "Late Night"

df["Time_Bucket"] = df["Hour"].apply(time_bucket)
```

```
print("Total Revenue:", round(df["Revenue"].sum(), 2))
print("Total Transactions:", df["transaction_id"].nunique())
print("Total Quantity:", df["transaction_qty"].sum())
print("Average Transaction Value:",
      round(df.groupby("transaction_id")["Revenue"].sum().mean(), 2))
```

```
hourly_sales = df.groupby("Hour")["Revenue"].sum()
hourly_sales
```

```
plt.figure(figsize=(10,5))
plt.plot(hourly_sales.index, hourly_sales.values, marker="o")
plt.title("Hourly Revenue")
plt.xlabel("Hour")
plt.ylabel("Revenue")
plt.show()
```

```
store_sales = df.groupby("store_location")["Revenue"].sum()
store_sales
```

```
plt.figure(figsize=(8,5))
store_sales.plot(kind="bar")
plt.title("Revenue by Store")
plt.xlabel("Store Location")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```
category_sales = df.groupby("product_category")["Revenue"].sum()
category_sales
```

```
plt.figure(figsize=(10,5))
category_sales.plot(kind="bar")
plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```
top_products = (
    df.groupby("product_detail")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
top_products
```

```
plt.figure(figsize=(10,6))
top_products.sort_values().plot(kind="barh")
plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.show()
```

```
time_sales = df.groupby("Time_Bucket")["Revenue"].sum()
time_sales
```

```
plt.figure(figsize=(8,5))
time_sales.plot(kind="bar")
plt.title("Revenue by Time of Day")
plt.xlabel("Time Bucket")
plt.ylabel("Revenue")
plt.show()
```

```
heatmap_data = df.pivot_table(
    values="Revenue",
    index="store_location",
    columns="Hour",
    aggfunc="sum"
)

plt.figure(figsize=(14,6))
sns.heatmap(heatmap_data)
plt.title("Store-wise Hourly Revenue")
plt.show()
```

```
peak_hour = df.groupby("Hour")["transaction_id"].count().idxmax()
best_store = df.groupby("store_location")["Revenue"].sum().idxmax()
best_category = df.groupby("product_category")["Revenue"].sum().idxmax()

print("Peak Transaction Hour:", peak_hour)
print("Best Performing Store:", best_store)
print("Best Product Category:", best_category)
```

## Conclusion

The analysis identifies hourly demand, store performance, product category performance and top products. The supplied dataset contains time-of-day but no transaction date, so daily/monthly/weekly calendar trends are not calculated.
